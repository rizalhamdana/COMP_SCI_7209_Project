```python
def join_all_year_daily_weather_data(yearly_data = []):
    all_years_dataframe = []
    
    for year in yearly_data:
        for _, weather_data in year.items():
            all_years_dataframe.append(weather_data)
    
    return pd.concat(all_years_dataframe, ignore_index=True)
```

```python
all_years_montly_weather_data_selected_states = join_all_year_daily_weather_data(yearly_data=[
    monthly_weather_data_states_2017, 
    monthly_weather_data_states_2018, 
    monthly_weather_data_states_2019,
    monthly_weather_data_states_2020, 
    monthly_weather_data_states_2021, 
    monthly_weather_data_states_2022
])
```

```python

corn_counties = set(selected_states_corn_production[['state_name', 'county_name']].apply(lambda x: (x[0], x[1]), axis=1))
cotton_counties = set(selected_states_cotton_production[['state_name', 'county_name']].apply(lambda x: (x[0], x[1]), axis=1))
wheat_counties = set(selected_states_wheat_production[['state_name', 'county_name']].apply(lambda x: (x[0], x[1]), axis=1))
soybean_counties = set(selected_states_soybeans_production[['state_name', 'county_name']].apply(lambda x: (x[0], x[1]), axis=1))

weather_counties = set(all_years_montly_weather_data_selected_states[['State', 'County']].apply(lambda x: (x[0], x[1]), axis=1))

print("County Matching Analysis:")
print("=" * 50)

# Check each crop
for crop_name, crop_counties in [('Corn', corn_counties), ('Cotton', cotton_counties), 
                                 ('Wheat', wheat_counties), ('Soybeans', soybean_counties)]:
    
    matched = crop_counties.intersection(weather_counties)
    crop_only = crop_counties - weather_counties
    
    print(f"\n{crop_name}:")
    print(f"  Crop counties: {len(crop_counties)}")
    print(f"  Matched with weather: {len(matched)}")
    print(f"  Missing weather data: {len(crop_only)}")
    print(f"  Coverage rate: {len(matched)/len(crop_counties)*100:.1f}%")
    
    if crop_only:
        print(f"  Counties missing weather: {list(crop_only)[:5]}...")  # Show first 5
```
